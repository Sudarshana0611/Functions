# 1. What is the difference between a function and a method in Python?
ans . A function is a block of code defined using def and can be called independently.

A method is a function associated with an object (like a string, list, or class instance) and is called using dot notation (e.g., list.append()).

Example:

Function: len([1, 2, 3])

Method: [1, 2, 3].append(4)

```python
def add(a, b):
    return a + b

result = add(2, 3)  # Calling the function
print(result)  # Output: 5
```

```python
class Calculator:
    def add(self, a, b):  # 'add' is a method
        return a + b

calc = Calculator()
result = calc.add(2, 3)  # Calling the method on an instance
print(result)  # Output: 5
```

5. What are iterators in Python and how do they differ from iterables?
ANS - Iterator:
An iterator is an object that keeps track of its position during iteration.
Has two methods: __iter__() and __next__()
Retrieves one item at a time using next()
Raises StopIteration when items are exhausted

Iterable:
An iterable is any Python object you can loop over (e.g., using a for loop).
Examples: list, tuple, str, dict, set
Has a __iter__() method that returns an iterator.

```python
# Iterable
nums = [1, 2, 3]

# Iterator
it = iter(nums)

print(next(it))  # 1
print(next(it))  # 2
```

6. Explain the concept of generators in Python and how they are defined.
ANS- Generators in Python
A generator is a special type of iterator that generates values on the fly using the yield keyword, rather than storing all values in memory at once.

Key Features:
Memory-efficient: Generates one item at a time.
Lazy evaluation: Doesn't compute values until needed.
Automatically maintains state between yield calls.
How to Define a Generator
Using a function with yield:



```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1
```

7. What are the advantages of using generators over regular functions?
ANS- I. Memory Efficiency : Generators yield one item at a time, avoiding memory-heavy data structures like lists.
Useful for large datasets or infinite sequences.

II. Lazy Evaluation : Values are produced only when needed, improving performance in certain cases.

III. Simpler Code for Iteration : Generators automatically manage state between calls, reducing the need for manual indexing or loop logic.

IV. Infinite Sequences Support : You can model endless data streams (like sensor data or number sequences) without running out of memory.

V. Improved Performance : For large computations or streaming data, generators are often faster due to reduced overhead.

VI. Cleaner Syntax with yield : Easier to write and understand compared to custom iterator classes.

```python
def squares(n):
    for i in range(n):
        yield i * i
```

9. Explain the purpose and usage of the `map()` function in Python.
ANS - The map() function is used to apply a function to every item in an iterable (like a list, tuple, etc.) and return a map object (an iterator).
SYNTAX: map(function, iterable)


```python
nums = [1, 2, 3, 4]
squares = map(lambda x: x * x, nums)
print(list(squares))  # Output: [1, 4, 9, 16]
```

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
ANS - map() Function
Purpose: Applies a given function to each item in an iterable.
Output: Returns a new iterator with the transformed values.
Example use case: Converting temperatures, squaring numbers, etc.

filter() Function
Purpose: Applies a function to test each item and filters out those for which the function returns False.
Output: Returns an iterator with only the items that satisfy the condition.
Example use case: Filtering even numbers, valid emails, etc.

reduce() Function
Purpose: Applies a function cumulatively to the items of an iterable, reducing the iterable to a single value.
Output: Returns a single aggregated result.
Example use case: Summing a list, computing factorial, finding maximum.

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    if n < 2 or n > 200:
        return False  # Not in the valid range or not prime
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(7))
print(is_prime(150))
print(is_prime(201))


```

```python
#  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
#terms.
class Fibonacci:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.a

fib = Fibonacci(10)
for num in fib:
    print(num, end=" ")






```

```python
# Write a generator function in Python that yields the powers of 2 up to a given
# exponent.
def power_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

for power in power_of_two(5):
    print(power)



```

```python
# Create a Python program that uses `filter()` to remove all the vowels from a given string.
def remove_vowels(text):
    vowels = "aeiouAEIOU"
    # Use filter to keep only characters that are not vowels
    result = filter(lambda char: char not in vowels, text)
    # Convert the filter object back to a string
    return "".join(result)

# Example usage
my_string = "Hello, World!"
filtered_string = remove_vowels(my_string)
print(filtered_string)  # Output: Hll, Wrld!
```
